```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
```

```python
arquivo = files.upload()
```

```python
img = cv2.imread('guara.jpg')
```

```python
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

```python
plt.figure(figsize=(5,3))
plt.imshow(img_gray, cmap='gray')
plt.title('Imagem em Cinza')
plt.axis('off')
```

```python
img_1bit = np.round(img_gray / 255) * 255

img_2bits = np.round(img_gray / 255 * 3) / 3 * 255

img_4bits = np.round(img_gray / 255 * 15) / 15 * 255

img_8bits = np.round(img_gray / 255 * 255)
```

```python
fig, ax = plt.subplots(1, 5, figsize=(15, 4))

ax[0].imshow(img_gray, cmap='gray')
ax[0].set_title('Original')
ax[0].axis('off')

ax[1].imshow(img_1bit, cmap='gray')
ax[1].set_title('1 bit - 2 níveis')
ax[1].axis('off')

ax[2].imshow(img_2bits, cmap='gray')
ax[2].set_title('2 bits - 4 níveis')
ax[2].axis('off')

ax[3].imshow(img_4bits, cmap='gray')
ax[3].set_title('4 bits - 16 níveis')
ax[3].axis('off')

ax[4].imshow(img_8bits, cmap='gray')
ax[4].set_title('8 bits - 256 níveis')
ax[4].axis('off')
```

```python
plt.tight_layout()
plt.savefig('quantizacao.png')
plt.show()
```

#O que acontece com os detalhes e com as transições de intensidade da imagem à medida que reduzimos o número de bits?

A imagem perde detalhes e as transições de intensidade ficam menos suaves e mais bruscas.
